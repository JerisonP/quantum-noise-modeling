# Rotating-Frame Hamiltonian and Density Matrix Evolution

This document provides a complete derivation of the rotating-frame Hamiltonian transformation and the time evolution of the density matrix, including all mathematical steps.

---

## Overview

We begin with the density matrix $\rho$ evolving under the lab-frame Hamiltonian $H_{\text{lab}}$ according to the Lindblad equation:

$$
\frac{d\rho}{dt} = -i[H_{\text{lab}}, \rho]
$$

The lab-frame Hamiltonian is given by:

$$
H_{\text{lab}} = \frac{1}{2} (\omega_q + \zeta(t)) \sigma_z + f(t) \cos(\omega_d t) \sigma_x
$$

**Parameters:**
- $\omega_q$ — qubit frequency
- $\zeta(t)$ — time-dependent noise
- $f(t)$ — drive amplitude
- $\omega_d$ — drive frequency

To simplify our analysis, we transform into a rotating frame that oscillates at the drive frequency $\omega_d$ using the unitary operator:

$$
S(t) = \exp\left(-i \frac{\omega_d t}{2} \sigma_z\right)
$$

The rotating-frame Hamiltonian is then:

$$
H_{\text{R.F.}} = S^{\dagger}(t) H_{\text{lab}} S(t) - i S^{\dagger}(t) \frac{dS(t)}{dt}
$$

---

## Derivation of the Rotating-Frame Hamiltonian

### Step 1: Transform the $\sigma_z$ Term

The first term in $H_{\text{lab}}$ commutes with $S(t)$ since both are diagonal in the same basis:

$$
S^{\dagger} \left( \frac{1}{2} (\omega_q + \zeta(t)) \sigma_z \right) S = \frac{1}{2} (\omega_q + \zeta(t)) \sigma_z
$$

This term remains unchanged under the rotation.

---

### Step 2: Transform the $\sigma_x$ Term

The drive term does not commute with $S(t)$. We use the rotation identity:

$$
S^{\dagger}(t) \sigma_x S(t) = \cos(\omega_d t) \sigma_x + \sin(\omega_d t) \sigma_y
$$

Applying this to the drive term:

$$
S^{\dagger}(t) \left( f(t) \cos(\omega_d t) \sigma_x \right) S(t) = f(t) \cos(\omega_d t) \left( \cos(\omega_d t) \sigma_x + \sin(\omega_d t) \sigma_y \right)
$$

Using trigonometric identities $\cos^2(\theta) = \frac{1 + \cos(2\theta)}{2}$ and $\cos(\theta)\sin(\theta) = \frac{\sin(2\theta)}{2}$:

$$
S^{\dagger}(t) \left( f(t) \cos(\omega_d t) \sigma_x \right) S(t) = f(t) \left( \frac{1 + \cos(2 \omega_d t)}{2} \sigma_x + \frac{\sin(2 \omega_d t)}{2} \sigma_y \right)
$$

---

### Step 3: Compute the Non-Inertial Term

The transformation also introduces a non-inertial correction term. We calculate:

$$
\frac{dS(t)}{dt} = \frac{d}{dt} \exp\left(-i \frac{\omega_d t}{2} \sigma_z\right) = -i \frac{\omega_d}{2} \sigma_z S(t)
$$

Therefore:

$$
-i S^{\dagger}(t) \frac{dS(t)}{dt} = -i S^{\dagger}(t) \left( -i \frac{\omega_d}{2} \sigma_z S(t) \right) = -\frac{\omega_d}{2} \sigma_z
$$

---

### Step 4: Assemble the Rotating-Frame Hamiltonian

Combining all transformed terms:

$$
H_{\text{R.F.}} = \frac{1}{2} (\omega_q + \zeta(t)) \sigma_z - \frac{\omega_d}{2} \sigma_z + f(t) \left( \frac{1 + \cos(2 \omega_d t)}{2} \sigma_x + \frac{\sin(2 \omega_d t)}{2} \sigma_y \right)
$$

Simplifying:

$$
H_{\text{R.F.}} = \frac{1}{2} (\omega_q - \omega_d + \zeta(t)) \sigma_z + f(t) \left( \frac{1 + \cos(2 \omega_d t)}{2} \sigma_x + \frac{\sin(2 \omega_d t)}{2} \sigma_y \right)
$$

---

### Step 5: Apply the Rotating Wave Approximation

Under the **Rotating Wave Approximation (RWA)**, we neglect rapidly oscillating terms at frequency $2\omega_d$, leaving:

$$
H_{\text{R.F.}} = \frac{1}{2} (\Delta + \zeta(t)) \sigma_z + \frac{f(t)}{2} \sigma_x
$$

where the **detuning** is defined as:

$$
\Delta = \omega_q - \omega_d
$$

---

## Time Evolution of the Density Matrix

The density matrix evolves in the rotating frame according to:

$$
\frac{d\rho}{dt} = -i [H_{\text{R.F.}}, \rho]
$$

### Matrix Representations

Express the Hamiltonian and density matrix explicitly:

$$
H_{\text{R.F.}} = \begin{pmatrix} 
\frac{1}{2} (\Delta + \zeta(t)) & \frac{f(t)}{2} \\ 
\frac{f(t)}{2} & -\frac{1}{2} (\Delta + \zeta(t)) 
\end{pmatrix}
$$

$$
\rho = \begin{pmatrix} 
\rho_{00} & \rho_{01} \\ 
\rho_{10} & \rho_{11} 
\end{pmatrix}
$$

---

### Computing the Commutator

The commutator $[H_{\text{R.F.}}, \rho] = H_{\text{R.F.}} \rho - \rho H_{\text{R.F.}}$ yields:

$$
[H_{\text{R.F.}}, \rho] = \begin{pmatrix} 
\frac{f(t)}{2} (\rho_{10} - \rho_{01}) & (\Delta + \zeta(t)) \rho_{01} - \frac{f(t)}{2} (\rho_{11} - \rho_{00}) \\ 
-(\Delta + \zeta(t)) \rho_{10} + \frac{f(t)}{2} (\rho_{11} - \rho_{00}) & \frac{f(t)}{2} (\rho_{01} - \rho_{10}) 
\end{pmatrix}
$$

---

### Final Evolution Equation

The time derivative of the density matrix is:

$$
\frac{d\rho}{dt} = -i \begin{pmatrix} 
\frac{f(t)}{2} (\rho_{10} - \rho_{01}) & (\Delta + \zeta(t)) \rho_{01} - \frac{f(t)}{2} (\rho_{11} - \rho_{00}) \\ 
-(\Delta + \zeta(t)) \rho_{10} + \frac{f(t)}{2} (\rho_{11} - \rho_{00}) & \frac{f(t)}{2} (\rho_{01} - \rho_{10}) 
\end{pmatrix}
$$

This provides the complete time evolution of all density matrix elements in the rotating frame under the influence of both the drive field and time-dependent noise.

---

## Summary

We have derived:

1. The **rotating-frame Hamiltonian** by applying a unitary transformation to remove explicit time dependence from the drive term
2. The **simplified Hamiltonian** under the RWA, valid when $\omega_d \approx \omega_q$
3. The **explicit time evolution equations** for all density matrix elements

This framework is essential for understanding driven qubit dynamics in the presence of noise and decoherence.

```julia
"""
Demo: Interaction Picture + Superoperator with DRIVE AMPLITUDE NOISE

This notebook demonstrates the interaction picture approach with
MULTIPLICATIVE NOISE on the drive amplitude.

Noise model: H(t) = f(t)*(1 + ζ(t))/2 * σx
where ζ(t) is Ornstein-Uhlenbeck noise

"""

using Plots
using LinearAlgebra
using Printf

# Load the library
include("interaction_picture_formalism.jl")
```

```julia
println("\n" * "="^70)
println(" Interaction Picture Superoperator Demo")
println(" (Drive Amplitude Noise Model)")
println("="^70 * "\n")

# ============================================================================
# EXAMPLE 1: Basic Usage
# ============================================================================

println("EXAMPLE 1: Basic Usage - π/2 Rotation with Drive Noise")
println("-" ^70)

# System parameters
tg = 1.0        # Gate time
θ = π/2         # Target rotation angle (π/2 gate)

# Noise parameters (multiplicative noise on drive)
τ_c = 0.1       # Correlation time
σ = 0.1         # Noise amplitude
n_samples = 50  # Number of noise realizations

# STEP 1: Compute the interaction picture superoperator (do this ONCE)
println("\nStep 1: Computing averaged superoperator...")
sup_I = compute_interaction_picture_superoperator(
    tg, θ, τ_c, σ, n_samples, 
    n_times=501, verbose=true
)

# STEP 2: Apply to initial state (can do this many times!)
println("\nStep 2: Applying to ground state |0⟩...")
ρ0 = ground_state()
result = apply_interaction_superoperator(ρ0, sup_I, verbose=true)

# Extract populations
pops = extract_populations(result.ρ_lab)

# Plot
p1 = plot(result.times, pops.P0, 
         label="P₀ (ground)", linewidth=2,
         xlabel="Time", ylabel="Population",
         title="Interaction Picture Evolution (Drive Noise)",
         legend=:right)
plot!(p1, result.times, pops.P1, 
      label="P₁ (excited)", linewidth=2)

display(p1)
savefig(p1, "example1_drive_noise.png")

# Final state
ρ_final = result.ρ_lab[end]
println("\nFinal state:")
println("  ρ₀₀ = ", @sprintf("%.6f", real(ρ_final[1,1])))
println("  ρ₁₁ = ", @sprintf("%.6f", real(ρ_final[2,2])))
println("  Purity = ", @sprintf("%.6f", purity(ρ_final)))

# Ideal π/2 gate would give P₀ = P₁ = 0.5
println("\nIdeal π/2 gate: P₀ = P₁ = 0.5")
println("Our result: P₀ = ", @sprintf("%.3f", pops.P0[end]), 
        ", P₁ = ", @sprintf("%.3f", pops.P1[end]))
println("Error: ", @sprintf("%.3f%%", 100*abs(pops.P1[end] - 0.5)/0.5))

println("\n" * "="^70 * "\n")
```

```julia
# ============================================================================
# EXAMPLE 2: Comparison - Lab Frame vs Interaction Picture
# ============================================================================

println("EXAMPLE 2: Lab Frame vs Interaction Picture")
println("-" ^70)
println("Both methods should give identical results!\n")

# Compute lab frame superoperator
println("Computing lab frame superoperator...")
sup_lab = compute_lab_frame_superoperator(
    tg, θ, τ_c, σ, n_samples,
    n_times=501, verbose=true
)

# Apply both methods
println("\nApplying interaction picture...")
result_I = apply_interaction_superoperator(ρ0, sup_I, verbose=false)

println("Applying lab frame...")
result_lab = apply_lab_superoperator(ρ0, sup_lab, verbose=false)

# Extract populations
pops_I = extract_populations(result_I.ρ_lab)
pops_lab = extract_populations(result_lab.ρ_lab)

# Plot comparison
p2 = plot(result_I.times, pops_I.P0,
         label="Interaction: P₀", linewidth=3, linestyle=:solid,
         xlabel="Time", ylabel="Population",
         title="Lab vs Interaction Picture (Drive Noise)")
plot!(p2, result_I.times, pops_I.P1,
      label="Interaction: P₁", linewidth=3, linestyle=:solid)
plot!(p2, result_lab.times, pops_lab.P0,
      label="Lab: P₀", linewidth=2, linestyle=:dash, alpha=0.7)
plot!(p2, result_lab.times, pops_lab.P1,
      label="Lab: P₁", linewidth=2, linestyle=:dash, alpha=0.7)

display(p2)
savefig(p2, "example2_comparison_drive_noise.png")

# Check difference
diff = norm(result_I.ρ_lab[end] - result_lab.ρ_lab[end])
println("\nFinal state difference: ", @sprintf("%.2e", diff))
println("(Should be very small - numerical error only)")

println("\n" * "="^70 * "\n")
```

```julia
# ============================================================================
# EXAMPLE 3: Multiple Initial States
# ============================================================================

println("EXAMPLE 3: Multiple Initial States with Same Superoperator")
println("-" ^70)
println("Compute superoperator ONCE, apply to MANY states!\n")

# Use the same superoperator we already computed
println("Using previously computed superoperator...")
println("Applying to three different initial states:\n")

# Initial states
ρ0_ground = ground_state()
ρ0_excited = excited_state()
ρ0_super = superposition_state(1.0, 1.0)  # (|0⟩ + |1⟩)/√2

# Evolve all three
println("1. Ground state |0⟩")
result_ground = apply_interaction_superoperator(ρ0_ground, sup_I, verbose=false)

println("2. Excited state |1⟩")
result_excited = apply_interaction_superoperator(ρ0_excited, sup_I, verbose=false)

println("3. Superposition (|0⟩+|1⟩)/√2")
result_super = apply_interaction_superoperator(ρ0_super, sup_I, verbose=false)

# Extract populations
pops_g = extract_populations(result_ground.ρ_lab)
pops_e = extract_populations(result_excited.ρ_lab)
pops_s = extract_populations(result_super.ρ_lab)

# Plot all three
p3a = plot(result_ground.times, pops_g.P0,
          label="P₀", linewidth=2,
          xlabel="Time", ylabel="Population",
          title="Initial: |0⟩", ylims=(0,1))
plot!(p3a, result_ground.times, pops_g.P1,
      label="P₁", linewidth=2)

p3b = plot(result_excited.times, pops_e.P0,
          label="P₀", linewidth=2,
          xlabel="Time", ylabel="Population",
          title="Initial: |1⟩", ylims=(0,1))
plot!(p3b, result_excited.times, pops_e.P1,
      label="P₁", linewidth=2)

p3c = plot(result_super.times, pops_s.P0,
          label="P₀", linewidth=2,
          xlabel="Time", ylabel="Population",
          title="Initial: (|0⟩+|1⟩)/√2", ylims=(0,1))
plot!(p3c, result_super.times, pops_s.P1,
      label="P₁", linewidth=2)

p3 = plot(p3a, p3b, p3c, layout=(3,1), size=(800, 1000))
display(p3)
savefig(p3, "example3_multiple_states.png")

println("\n✓ All three states evolved using the SAME superoperator!")

println("\n" * "="^70 * "\n")
```

```julia
# ============================================================================
# EXAMPLE 4: Effect of Noise Strength
# ============================================================================
println("EXAMPLE 4: Effect of Noise Strength")
println("-" ^70)
println("Comparing different noise amplitudes\n")

σ_values = [0.05, 0.1, 0.2]
results_noise = []

for σ_test in σ_values
    println("Computing for σ = $σ_test...")
    sup_test = compute_interaction_picture_superoperator(
        tg, θ, τ_c, σ_test, 30,
        n_times=301, verbose=false
    )
    res = apply_interaction_superoperator(ρ0, sup_test, verbose=false)
    push!(results_noise, res)
end

# Plot BOTH P₀ and P₁
p4 = plot(xlabel="Time", ylabel="Population",
         title="Effect of Drive Noise Amplitude",
         legend=:bottomleft)  # Bottom left corner

# Plot P₀ (ground state) with solid lines
for (i, σ_test) in enumerate(σ_values)
    pops_test = extract_populations(results_noise[i].ρ_lab)
    plot!(p4, results_noise[i].times, pops_test.P0,
          label="P₀, σ = $σ_test", linewidth=2, linestyle=:solid)
end

# Plot P₁ (excited state) with dashed lines
for (i, σ_test) in enumerate(σ_values)
    pops_test = extract_populations(results_noise[i].ρ_lab)
    plot!(p4, results_noise[i].times, pops_test.P1,
          label="P₁, σ = $σ_test", linewidth=2, linestyle=:dash)
end

hline!(p4, [0.5], label="Ideal π/2", linestyle=:dot, color=:black, 
       alpha=0.5, linewidth=2)

display(p4)
savefig(p4, "example4_noise_strength.png")

println("\nFinal populations:")
for (i, σ_test) in enumerate(σ_values)
    pops_test = extract_populations(results_noise[i].ρ_lab)
    println("  σ = $σ_test: P₀ = ", @sprintf("%.4f", pops_test.P0[end]), 
            ", P₁ = ", @sprintf("%.4f", pops_test.P1[end]))
end
println("\n✓ Stronger noise → worse gate performance")

println("\n" * "="^70 * "\n")
```
